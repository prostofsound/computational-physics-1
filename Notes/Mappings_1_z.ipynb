## Computational exploration of Mappings by 1/z

```python
# Import necessary libraries
import numpy as np
import matplotlib.pyplot as plt
import scienceplots
```

```python
# Set plots style
plt.style.use(['science', 'notebook', 'no-latex'])
```

```python
def transformation_1_z(x, y, boolean, name):
    """
    This function computes the transformation 1/z of a given complex number and then plots the map in the w-plane.
    ___________________________________________
    Inputs:
    x -> array, real part of z.
    y -> array, imaginary part of z.
    boolean -> True or False, to save the image.
    name -> str, title of the plot
    ___________________________________________
    Output:
    map of the transformation in the w-plane.
    """
    #Compute the transformation
    denom = (x**2 + y**2)
    u = x/denom
    v = -y/denom

    #Plotting
    plt.figure(figsize = (5,4))
    plt.plot(u, v, color = "crimson", linestyle = "-")
    plt.title(f"Mapping 1/z of a {name}")
    plt.xlabel(r"$u$")
    plt.ylabel(r"$v$")
    # Marking the Cartesian plane
    plt.axhline(0, color='black', linewidth=1, linestyle = "--")  # x-axis (v=0)
    plt.axvline(0, color='black', linewidth=1, linestyle = "--")  # y-axis (u=0)
    plt.grid(True, linestyle=':', alpha=0.7)
    if boolean:
        plt.savefig(f"{name}.pdf", dpi=300)
    plt.show()
```

### Example 1: Vertical line
$$
x = c_1, c_1 \ne 0
$$

```python
#Generate the y vector
y = np.linspace(-15, 15, 1000)
#Define x
x = -0.5 #negative
#Mapping
transformation_1_z(x, y, True, "negative vertical line")
```

```python
#Define x
x = 0.7 #positive
#Mapping
transformation_1_z(x, y, True, "positive vertical line")
```

### Example 2: Horizontal line

```python
#Generate the x vector
x = np.linspace(-15, 15, 1000)
#Define y
y = -0.5 #negative
#Mapping
transformation_1_z(x, y, True , "negative horizontal line")
```

### Example 3: Circle
$$
|z - (1 + i)| = 1
$$
We can express the circle as,
$$
|(x - 1) - i(y - 1)| = 1,
$$
$$
\sqrt{(x - 1)² + (y - 1)²} = 1,
$$
$$
(x - 1)² + (y - 1)² = 1,
$$
$$
y = 1 \pm\sqrt{1 - (x - 1)²}.
$$

```python
#Generate the x vector
x = np.linspace(0, 2, 500)
#Define the circle in terms of y.
y1 = 1 + np.sqrt(1 - (x - 1)**2) #positive
y2 = 1 - np.sqrt(1 - (x - 1)**2) #negative
# Concatenate x and y for full circle
x_full = np.concatenate([x, x[::-1]])
y_full = np.concatenate([y1, y2[::-1]])
#print(y_full)
#Mapping
transformation_1_z(x_full, y_full, True , "Circle")
```
