# Extrema of functions

We can find the extrema (minima and maxima) of functions using scipy.

In this case, we use its signal procesing routines in scipy.signal.

## Example:

Find the maxima and minima of the function given by the following data:

https://github.com/wbandabarragan/physics-teaching-data/blob/main/1D-data/cooltable.dat

"cooltable.dat", which has two columns:

1. Temperatures of the gas in units of: $\rm K$.

2. Cooling rates (how much energy is being losed per unit time per unit number density) in units of: $\rm erg\,cm^3\,s^{-1}$.


We want to identify the minima and maxima of the cooling function.

An atomic gas contains a lot of hydrogen and helium.

```python
import matplotlib.pyplot as plt

import pandas as pd

import numpy as np

import scipy.signal as ss
```

### Read table:

```python
df = pd.read_csv("/home/prostofsound/jnotebooks/computational-physics-1/Notes/data/cooltable.dat", sep = '\s+', header = None)
```

```python
print(df)
```

```python
# Linear scale plot

plt.figure(figsize=(5,4))

plt.plot(df[0], df[1])

plt.show()
```

Both axes need a logarithmic scale, the orders of magnitue difference is very high.

```python
temperatures = np.log10(np.array(df[0]))

coolingrates = np.log10(np.array(df[1]))
#print(type(temperatures))
```

```python
print(coolingrates)
```

```python
#Log scale plot

fig, ax = plt.subplots(figsize=(10, 4))

plt.plot(temperatures, coolingrates)

plt.xlabel(r"$T\,[K]$")

plt.ylabel(r"$j\,[erg\,cm^3\,s^{-1}$")

plt.xlim(1,10)
plt.ylim(-26, -21)

plt.show()
```

At 1e6 K, most of the atomic gas has already been ionized.

The first peak is the emission of H, which is expected because of quantum mechanical transitions.

If we'd like to check for molecular gas, the temperature regime where they can exist is 1e3 K max.

## Peak finding:

We can use find_peaks().

https://docs.scipy.org/doc/scipy/reference/generated/scipy.signal.find_peaks.html

```python
peaks, _ = ss.find_peaks(coolingrates)
```

```python
print(peaks)
```

```python
# Actual extrema
t_ext = temperatures[peaks]
print(t_ext)

c_ext = coolingrates[peaks]
print(c_ext)
```

```python
fig, ax = plt.subplots(figsize=(10, 4))

plt.plot(temperatures, coolingrates)
plt.plot(temperatures[peaks], coolingrates[peaks], "o")

plt.xlabel(r"$T\,[K]$")

plt.ylabel(r"$j\,[erg\,cm^3\,s^{-1}$")

#plt.xlim(4,10)
#plt.ylim(-24, -21)

plt.show()
```

## Extrema finding:
https://docs.scipy.org/doc/scipy/reference/generated/scipy.signal.argrelextrema.html

To find the relative extrema.

```python
maxima = ss.argrelextrema(coolingrates, np.greater)

minima = ss.argrelextrema(coolingrates, np.less)
```

```python
print(maxima)
```

```python
print(minima)
```

```python
extrema = np.concatenate([maxima, minima], axis=None)

print(extrema)
```

```python
fig, ax = plt.subplots(figsize=(10, 4))


plt.plot(temperatures, coolingrates)
plt.plot(temperatures[extrema], coolingrates[extrema], "o")

plt.show()
```

```python
fig, ax = plt.subplots(figsize=(10, 4))

plt.plot(temperatures, coolingrates)
plt.plot(temperatures[maxima], coolingrates[maxima], "o", label = 'maxima')
plt.plot(temperatures[minima], coolingrates[minima], "d", label = 'minima')
plt.legend()
plt.show()
```

```python
fig, ax = plt.subplots(figsize=(10, 4))

plt.plot(temperatures, coolingrates, linestyle = " ", marker = '+')
plt.plot(temperatures[maxima], coolingrates[maxima], "o", label = 'maxima')
plt.plot(temperatures[minima], coolingrates[minima], "d", label = 'minima')

plt.xlim(6.5,7.5)
plt.ylim(-22.8,-22.2)
plt.legend()
plt.show()
```

Some extrema are still missing because it could happen that we have two equal values in the y-axis.

## Extrema finding with a different comparator:

```python
maxima2 = ss.argrelextrema(coolingrates, np.greater_equal, mode = 'wrap') #this time we also check similar values

minima2 = ss.argrelextrema(coolingrates, np.less_equal, mode = 'wrap')
```

```python
extrema2 = np.concatenate([maxima2, minima2], axis=None)

print(extrema2)
```

```python
fig, ax = plt.subplots(figsize=(10, 4))

plt.plot(temperatures, coolingrates)
plt.plot(temperatures[maxima2], coolingrates[maxima2], "o", label = 'maxima')
plt.plot(temperatures[minima2], coolingrates[minima2], "d", label = 'minima')
plt.legend()
plt.show()
```

```python
fig, ax = plt.subplots(figsize=(10, 4))

plt.plot(temperatures, coolingrates, linestyle = " ", marker = '+')
plt.plot(temperatures[maxima2], coolingrates[maxima2], "o", label = 'maxima')
plt.plot(temperatures[minima2], coolingrates[minima2], "d", label = 'minima')

plt.xlim(6.5,7.5)
plt.ylim(-22.8,-22.2)
plt.legend()
plt.show()
```
